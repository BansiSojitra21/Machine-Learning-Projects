```python
X = df.drop(columns=["Outcome"])
y = df["Outcome"]

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=45)
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
lr = LogisticRegression()
lr.fit(X_train,y_train)
lrp = lr.predict(X_test)
print("Accuracy Score : ",accuracy_score(y_test,lrp))
```

```python
n_features = X_train.shape[1]

w = np.zeros(n_features)
b = 0

learning_rate = 0.01
epochs = 5000
```

```python
def sigmoid(z):
    return 1/(1 + np.exp(-z))
```

```python
def compute_loss(y,y_pred):

    m = len(y)
    epsilon = 1e-15
    
    y_pred = np.clip(y_pred,epsilon,1-epsilon)
        
    loss = -(1/m)*np.sum(y*np.log(y_pred)+(1-y)*np.log(1-y_pred))
    
    return loss
```

```python
def compute_gradient(X,y,y_pred):
    m = len(y)
    dw = (1/m) * np.dot(X.T,(y_pred-y))
    db = (1/m) * np.sum(y_pred-y)
    return dw,db
```

```python
loss_history = []

for epoch in range(epochs):

    # Linear Equation
    z = np.dot(X_train, w) + b

    # Prediction
    y_pred = sigmoid(z)

    # Loss
    loss = compute_loss(y_train, y_pred)

    loss_history.append(loss)

    # Gradient
    dw, db = compute_gradient(X_train, y_train, y_pred)

    # Update
    w = w - learning_rate * dw
    b = b - learning_rate * db

    if epoch % 100 == 0:
        print(f"Epoch {epoch}  Loss = {loss:.4f}")
```

```python
z_test = np.dot(X_test, w) + b

probability = sigmoid(z_test)

prediction = (probability >= 0.5).astype(int)
prediction
```

```python
accuracy = accuracy_score(y_test, prediction)

print("Accuracy :", accuracy)
```

```python
dt = DecisionTreeClassifier(random_state=56)
dt.fit(X_train,y_train)
dtp = dt.predict(X_test)
print("Accuracy Score : ",accuracy_score(y_test,dtp))
```

```python
rf = RandomForestClassifier(random_state=45,n_estimators=100,min_samples_leaf=1,min_samples_split=2)
rf.fit(X_train,y_train)
rfp = rf.predict(X_test)
print("Accuracy Score : ",accuracy_score(y_test,rfp))
```

```python
gb = GradientBoostingClassifier(n_estimators=100,learning_rate=0.01,max_depth=5,random_state=54)
gb.fit(X_train,y_train)
gbp = gb.predict(X_test)
accuracy_score(y_test,gbp)
```

```python
xgb = XGBClassifier(objective="binary:logistic",
    n_estimators=100,
    learning_rate=0.01,
    max_depth=4,
    random_state=34)
xgb.fit(X_train,y_train)
xgbp = xgb.predict(X_test)
accuracy_score(y_test,xgbp)
```
